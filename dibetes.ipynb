```python
import matplotlib.pyplot as plt

plt.scatter(df['Pregnancies'], df['Glucose'])  # fixed typo
plt.title('Pregnancies vs Glucose')
plt.xlabel('Pregnancies')
plt.ylabel('Glucose')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.scatter(df['Pregnancies'], df['Glucose'], color = 'blue', label='Glucose')
plt.scatter(df['Pregnancies'], df['BloodPressure'], color='red',label= 'Blood Pressure')
plt.xlabel('Pregnancies')
plt.ylabel('Values')
plt.title('Pregnancies vs Glucose & Blood Pressure')
plt.legend()
plt.show()
```

```python
mean_glucose = df['Glucose'].mean()
print("Average Glucose:", mean_glucose)

mean_preg = df['Pregnancies'].mean()
print("Average Pregnancies:", mean_preg)
```
